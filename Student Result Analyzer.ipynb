```python
print("Student Result Analyzer")
while True:
    name = input("Enter student name: ")
    
    m1 = int(input("Enter marks for Subject 1: "))
    m2 = int(input("Enter marks for Subject 2: "))
    m3 = int(input("Enter marks for Subject 3: "))

    total = m1 + m2 + m3
    percentage = total / 3

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 75:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    else:
        grade = "Fail"

    print("\n--- Result ---")
    print("Name:", name)
    print("Total Marks:", total)
    print("Percentage:", round(percentage, 2))
    print("Grade:", grade)

    choice = input("\nDo you want to enter another student? (yes/no): ").lower()
    if choice != "yes":
        print("Thankyou")
        print("PROJECT CREATED BY ABHISHEK GUPTA")
        break
```
